```python
LAST_NAME = "LUNTAO"
STUDENT_ID = "TUPM-26-1947"
FAVORITE_ARTIST = "RAKIM"

SEED_DIGIT = int(STUDENT_ID[-1])  # 7
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())  # 29
NAME_LENGTH = len(LAST_NAME)  # 6
ARTIST_LENGTH = len(FAVORITE_ARTIST)  # 5
ARTIST_SUM = sum(ord(c.upper()) - 64 for c in FAVORITE_ARTIST if c.isalpha())  # 52

print("=== STUDENT PARAMETERS ===")
print("Seed:", SEED_DIGIT | "ID Sum:", ID_SUM | "Name Length:", NAME_LENGTH)
print("Artist Length:", ARTIST_LENGTH | "Artist Sum:", ARTIST_SUM)
```

```python
# Generate datasets
original_records = [100 + ID_SUM + i * SEED_DIGIT for i in range(10)]
rotation_point = SEED_DIGIT % len(original_records)
rotated_records = original_records[rotation_point:] + original_records[:rotation_point]
target_incident = original_records[(ID_SUM + ARTIST_SUM) % len(original_records)]

def search_rotated(data, target):
    comparisons = 0
    for idx, item in enumerate(data):
        comparisons += 1
        if item == target:
            return idx, comparisons
    return -1, comparisons

found_idx, comps = search_rotated(rotated_records, target_incident)

print("=== EXERCISE 1: EMERGENCY DISPATCH SEARCH ===")
print("Original Dataset:", original_records)
print("Rotated Dataset:", rotated_records)
print("Target Incident:", target_incident)
print(f"Target Found at Index: {found_idx} with {comps} comparisons")
print("Time Complexity: O(n) | Space Complexity: O(1)")
```

```python
dept_a = sorted([1000 + SEED_DIGIT * i + ID_SUM for i in range(5)])
dept_b = sorted([2000 + ARTIST_LENGTH * i + NAME_LENGTH for i in range(5)])

def merge_sorted_records(arr1, arr2):
    consolidated = []
    i = j = comparisons = 0
    while i < len(arr1) and j < len(arr2):
        comparisons += 1
        if arr1[i] <= arr2[j]:
            consolidated.append(arr1[i])
            i += 1
        else:
            consolidated.append(arr2[j])
            j += 1
    consolidated.extend(arr1[i:])
    consolidated.extend(arr2[j:])
    return consolidated, comparisons

consolidated_list, merge_comps = merge_sorted_records(dept_a, dept_b)

print("=== EXERCISE 2: UNIVERSITY ENROLLMENT RECORDS ===")
print("Department A:", dept_a)
print("Department B:", dept_b)
print("Consolidated Dataset:", consolidated_list)
print("Records Before:", len(dept_a) + len(dept_b), "| After:", len(consolidated_list))
print("Comparisons Performed:", merge_comps)
print("Ordering Verification (Is Sorted?):", consolidated_list == sorted(consolidated_list))
print("Time Complexity: O(n + m) | Space Complexity: O(n + m)")
```

```python
base_orders = [101, 105, 102, 104, 103]

def bubble_sort(data):
    a, comps, shifts = data.copy(), 0, 0
    for i in range(len(a)):
        for j in range(len(a) - i - 1):
            comps += 1
            if a[j] > a[j+1]:
                a[j], a[j+1] = a[j+1], a[j]
                shifts += 1
    return a, comps, shifts

def selection_sort(data):
    a, comps, shifts = data.copy(), 0, 0
    for i in range(len(a) - 1):
        min_idx = i
        for j in range(i + 1, len(a)):
            comps += 1
            if a[j] < a[min_idx]:
                min_idx = j
        if min_idx != i:
            a[i], a[min_idx] = a[min_idx], a[i]
            shifts += 1
    return a, comps, shifts

def insertion_sort(data):
    a, comps, shifts = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comps += 1
            if a[j] <= key:
                break
            a[j+1] = a[j]
            shifts += 1
            j -= 1
        a[j+1] = key
    return a, comps, shifts

conditions = {
    "Sorted": sorted(base_orders),
    "Reverse": sorted(base_orders, reverse=True),
    "Random": base_orders
}

print("=== EXERCISE 3: E-COMMERCE ORDER PROCESSING ===")
for cond, dataset in conditions.items():
    print(f"\nCondition: {cond} -> {dataset}")
    for name, sort_fn in [("Bubble", bubble_sort), ("Selection", selection_sort), ("Insertion", insertion_sort)]:
        res, c, s = sort_fn(dataset)
        print(f"  [{name} Sort] Result: {res} | Comparisons: {c} | Movements/Swaps: {s}")
```
